```python
import requests

class AvatarAPI:
    BASE_URL = "https://last-airbender-api.fly.dev/api/v1/characters"

    def obter_todos_personagens(self, limite=50):
        url = f"{self.BASE_URL}?perPage={limite}"
        resposta = requests.get(url)
        if resposta.status_code == 200:
            return resposta.json()
        return []

    def buscar_por_nome(self, nome):
        url = f"{self.BASE_URL}?name={nome}"
        resposta = requests.get(url)
        if resposta.status_code == 200:
            return resposta.json()
        return []

    def buscar_por_afiliacao(self, afiliacao):
        url = f"{self.BASE_URL}?affiliation={afiliacao}"
        resposta = requests.get(url)
        if resposta.status_code == 200:
            return resposta.json()
        return []

class GerenciadorAvatar:
    def __init__(self):
        self.api = AvatarAPI()

    def exibir_personagem(self, p):
        nome = p.get('name', 'Desconhecido')
        afiliacao = p.get('affiliation', 'Sem afiliação registrada')
        aliados = ", ".join(p.get('allies', [])) or "Nenhum"
        inimigos = ", ".join(p.get('enemies', [])) or "Nenhum"

        print("=" * 50)
        print(f"Nome: {nome}")
        print(f"Afiliação: {afiliacao}")
        print(f"Aliados: {aliados}")
        print(f"Inimigos: {inimigos}")
        print("=" * 50)

    def executar(self):
        print("--- APLICAÇÃO AVATAR API ---")
        personagens = self.api.obter_todos_personagens(limite=5)
        print(f"\nExibindo os primeiros {len(personagens)} personagens:")
        for p in personagens:
            self.exibir_personagem(p)

        print("\nBuscando pelo personagem 'Aang':")
        resultados_aang = self.api.buscar_por_nome("Aang")
        for p in resultados_aang:
            self.exibir_personagem(p)

if __name__ == "__main__":
    app = GerenciadorAvatar()
    app.executar()
```
